# 1.1: Mean
- Sample Mean
- Population Mean

Calculating Mean

Manual Method

```python
population = [15,14,13,12,17,16]
sample = [1,9,2,4]

pop_mean = sum(population)/len(population)
print(pop_mean)

sample_mean = sum(sample)/len(sample)
print(sample_mean)
```

Using numpy

```python
import numpy as np

data = [15,14,13,12,17,16]
arithmatic_mean = np.mean(data)
print(arithmatic_mean)
```

Using Statistics

```python
import statistics as stat

var = [15,14,13,12,17,16]

mean_stat = stat.mean(var)
print(mean_stat)
```

on Real Dataset

```python
import seaborn as sns
import numpy as np

iris = sns.load_dataset("iris")
iris.head()
iris.shape

mean_sepal_length = np.mean(iris["sepal_length"])
print(mean_sepal_length)
```

# 1.2: Median

- The median is a measure of central tendency that represents the middle value of a dataset when it is arranged in ascending or descending order. 
- Unlike the mean, the median is not affected by outliers or skewed data, making it a robust measure for datasets with extreme values.

### Steps to calculate The Median

- Sort the dataset in ascending or descending order
- Determine the number of observations n
- Apply the appropriate formula depending on whether n is odd or even

Using numpy

```python
import numpy as np
data = [22,22,55,12,199,12,19]

median = np.median(data)
print(median)
```

Using Statistics

```python
import statistics as stat

data = [22,22,55,12,199,12,19]

median = stat.median(data)
print(median)
```

Using pandas

```python
import pandas as pd

data = [22,22,55,12,199,12,19]

df = pd.DataFrame({
    "values":data
})
median_pandas = df["values"].median()
print(median_pandas)
```

2: Measure of Variability - Range, Variance, Standard Deviation
- Also Known as Measure of Dispersion.
- Measure of Dispersion, statistics ma data k spread ya spread out hona k extend ko measure krta hy.
- ya batata hy k data points ak mean value say kitna dor hain.
- inka use data ky distribution ko samajhna or compare krna k liya hota hy.

- **For Example**: 
Agar hum sirf mean dekhta hain tu huma ya nahe pata chalta k data points mean se kitna dor hain, esi liya hum Measure of Dispersion ky help lata hain taka huma ya samjhna ma asani ho k data kitna spread out hy.

# 2.1: Range

- In Descriptive Statistics, range is a simple measure of variability or spread in a dataset.
- It indicates how far the data points are spread out by showing the difference between maximum and minimum values in a dataset

- **Key Points**

- Range gives a quick sense of dispersion: If the range is small, it indicates that the data points are close to each other. A large range suggest that the data points are more spread out.
- Range is sensitive to outliers: Since the range only looks at the two extreme values (Maximum and Minimum), any outlier can significantly increase the range, even if the rest of the data is closely packed.

# 2.2: Variance

- Variance ak statistical measure hy jo dataset ky values k beech ky variability ya dispersion ko quantify krta hy.
- Means that k , ya batata hy k data ky individual values mean say kitni dor hain, or ya dataset ka spread kitna hy.

- Population variance and Sample variance are measure of dispersion that quantify how much data points differ from the mean.

Using numpy

```python
import numpy as np

data = [2,5,6,6,7,5,9,14,10,2,3,4]
# agr hum log ddof nahe likh tb wo by default sample variance k liya nikalta hy
sample_variance = np.var(data,ddof=1) # ddof -> Delta Degree of Freedom

population_variance = np.var(data,ddof=0) 

print(sample_variance)
print(population_variance)
```

Using Pandas

```python
import pandas as pd

data = [2,5,6,6,7,5,9,14,10,2,3,4]

series = pd.Series(data)
sample_variance = series.var()
print(sample_variance)

population_variance = series.var(ddof=0)
print(population_variance)
```

Using scipy

```python
from scipy import stats

data = [2,5,6,6,7,5,9,14,10,2,3,4]

sample_variance = stats.tvar(data)
print(sample_variance)

sample_variance = stats.tvar(data,ddof=0)
print(sample_variance)
```
