# *Break, continue, and pass statements*

# break Statement

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

# continue Statement

```python
for i in range(10):
    if i == 2:
        continue
    print(i)
```

# pass Statement

```python
if True:
    pass
```
