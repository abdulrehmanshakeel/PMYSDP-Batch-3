# Missing Values/Data

- Jab kisi dataset me koi value available na ho to usay missing value kehte hain.

Pandas me missing value ko:

- NaN (Not a Number)

- None

se represent kiya jata hai.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Hina"],
    "Marks": [85, np.nan, 78, None],
    "City": ["Lahore", "Karachi", None, "Islamabad"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df.dropna(subset=["Marks"]) # agar specific column ky basis per krna ho tu
```

- axis=1 → column
- axis=0 → row

```python
df.dropna(axis=1)
```

# Filling Missing Values

### 1: Fixed Value Se Fill

```python
df["Marks"].fillna(0, inplace=True)
```

### 2: Mean Se Fill

```python
df["Marks"].fillna(df["Marks"].mean(), inplace=True)
```

### 3: Median Se Fill

```python
df["Marks"].fillna(df["Marks"].median(), inplace=True)
```

### 4: Mode Se Fill (Categorical Data)

```python
df["City"].fillna(df["City"].mode()[0], inplace=True)
```

### 5: Forward Fill (Previous Value Copy)

```python
# df.fillna(method="ffill", inplace=True)
df.ffill(inplace=True)
```

### 6: Backward Fill

```python
df.bfill(inplace=True)
```
