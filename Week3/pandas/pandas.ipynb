# Pandas ke 2 main data structures

# 1: Series

#### Series = 1D labeled array

```python
import pandas as pd

# Simple Series
s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
s = pd.Series([10, 20, 30], index=["a", "b", "c"])
print(s)
```

```python
print(s["a"])
```

# DataFrame

#### DataFrame = 2D table (rows + columns)

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [85, 90, 78],
    "City": ["Lahore", "Karachi", "Islamabad"]
}

df = pd.DataFrame(data)
print(df)
```
